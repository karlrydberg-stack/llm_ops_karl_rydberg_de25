```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv 

load_dotenv()

# chat completion style same as that of OpenAI, which Nvidia models support
model = OpenAIChatModel(
    "nvidia/nemotron-3.5-lightning-30b-a3b",  # any model from build.nvidia.com
    provider=OpenAIProvider( # connection details to nvidia
        base_url="https://integrate.api.nvidia.com/v1",
        api_key=os.environ["NVIDIA_API"],  

    ),
)

# wraps a model
chef_bot = Agent(
    model,
    system_prompt=(
        "You are a professional chef that is very good at making asian food and explaining in a pedagogical way how to create yummy dishes.",
        """
        Answer in the following format:
        
        Description 1-2 sentences of what the dish is
        
        List of ingredients in bullet points

        Short step by step description on how to make the dish
        """
        "Please output the result in a python list format, with each dish separated by a line break."
    ),
)
result = await chef_bot.run("Give me 3 asian soups in a list")
result
```

```python
print(result.output)
```

```python
with open("asian_soups.md", "w") as f:
    f.write(result.output)
```

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv
from pydantic_ai import ModelSettings 

load_dotenv()

# chat completion style same as that of OpenAI, which Nvidia models support
model = OpenAIChatModel(
    "nvidia/nemotron-3.5-lightning-30b-a3b",  # any model from build.nvidia.com
    provider=OpenAIProvider( # connection details to nvidia
        base_url="https://integrate.api.nvidia.com/v1",
        api_key=os.environ["NVIDIA_API"],  
    ),
)

agent = Agent(
    model,
    system_prompt="You are a children storyteller. Answer max 3 sentences.",
    model_settings=ModelSettings(thinking="low"),
)

prompt = "A gray netherland rabbit jumps ..."

result = await agent.run(prompt)
print(result.output)
```

```python
from pathlib import Path
from pydantic_ai.messages import BinaryContent

image_joker = Agent(
    model,
    system_prompt="Analyze the image and describe it in a few sentences. If you can, also provide a joke about the image.",
)

image_data = Path("kokchun.png").read_bytes()
prompt = "Tell me about his dude"

result = await image_joker.run(
    user_prompt = [prompt, BinaryContent(image_data, "image/png")]
)
```
